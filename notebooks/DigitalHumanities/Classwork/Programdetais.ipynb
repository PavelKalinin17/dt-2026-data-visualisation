```python
import pandas as plt
df= pd.read_csv("Competencies1.csv")
df.head() #Show the first few rows
df.describe() # Show some basic statistics
```

```python
import matplotlib.pyplot as plt
avg_hours = df.groupby("Student")["Hours_Studied"].mean()
#colours= plt.vm.plasms(range(len(avg_hours)))
avg_hours.plot(kind="bar", x="Student", y="Hours_Studies", color="#F54927")
plt.show()
               
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*50, alpha=0.6, color="#F54927")
plt.xlabel("Hours_studied")
plt.ylabel("Grades")
plt.title("Grade vs hours studies (Bubble size=hours)")
```

```python
plt.figure(figsize=(8,6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Distribution of study hours and grade")
```
